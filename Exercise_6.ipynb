```python
#checks whether the given number is pallindrome or not usimg function
def palindrome(num):
  original = num
  rev = 0
  while(num!=0):
    rem = num%10
    rev = rev*10+rem
    num = num//10
  if(rev == original):
    return True
  else:
    return False
print(palindrome(545))
print(palindrome(23345))
```

```python
#reverse of a string using fuction
def reverse_str(s):
  rev = ""
  for i in range(len(s)-1,-1,-1):
    rev=rev+s[i]
  return rev
print(reverse_str("abhi"))
```

```python
#example on default argument
def print_msg(name,branch = "AI&DS"):
  print("the name is",name,"the branch is",branch)
print_msg("shalini")
```

```python
#example on variable length
def sub(*num):
  print(num)
  print(sum(num))
sub(1,2,3,4,5,6)
```

```python
#function that returns multiple
def multiple_values(*num):
  s = sum(num)
  c = len(num)
  avg = s/c
  return s,c,avg
res_sum,res_count,res_avg=multiple_values(1,2,3,4,5)
print(f"sum is {res_sum},count is {res_count},the average is {res_avg}")
```
