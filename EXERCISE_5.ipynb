```python
#PALINDROME
def palindrome(n):
  original = n
  rev = 0
  while(n!=0):
    rem=n%10
    rev=rev*10+rem
    n=n//10
  if(rev==original):
    return True
  else:
    return False


print(palindrome(252))
print(palindrome(645))

```

```python
#multiple values in function
def mult_values(*num):
  s=sum(num)
  c=len(num)
  avg=s/c
  return s,c,avg
res_sum,res_count,res_avg=mult_values(1,2,3,4,5)
print(f"sum={res_sum},count={res_count},the average is {res_avg}")
```
